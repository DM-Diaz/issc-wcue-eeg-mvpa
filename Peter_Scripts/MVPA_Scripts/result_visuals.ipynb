```python
import os

import matplotlib.pyplot as plt
import numpy as np

from matplotlib.lines import Line2D
from scipy.io import savemat
```

```python
RESULTS_DIR = "/home/yuchin/Desktop/Sambashare/ISSP_mvpa/figures/results"
```

```python
times = np.linspace(-0.2, 0.796, 250)
```

```python
def plot_accuracy(accs, ax, title):
    mean_accs = np.mean(accs, axis=0)
    std = np.std(accs, axis=0) / 2

    ax.plot(times, mean_accs)
    ax.fill_between(
        times,
        mean_accs - std,
        mean_accs + std,
        color="gray",
        alpha=0.3,
        label="Standard Deviation",
    )

    ax.axhline(0.50, color="r", linestyle="--")
    ax.axvline(0, color="black")

    ax.set_ylim(0.4, 1)

    ax.set_title(title)
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Accuracy")
```

# Individual

```python
single_eegs_E1_train_accs = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E1_low-high_low-high_train_accs.npy")
)
single_eegs_E1_val_accs = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E1_low-high_low-high_val_accs.npy")
)
single_eegs_E1_weights = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E1_low-high_low-high_weights.npy")
)

single_eegs_E2_baseline_train_accs = np.load(
    os.path.join(
        RESULTS_DIR, "single_eegs_E2_baseline_low-high_low-high_train_accs.npy"
    )
)
single_eegs_E2_baseline_val_accs = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E2_baseline_low-high_low-high_val_accs.npy")
)
single_eegs_E2_baseline_weights = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E2_baseline_low-high_low-high_weights.npy")
)

single_eegs_E2_penalty_train_accs = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E2_penalty_low-high_low-high_train_accs.npy")
)
single_eegs_E2_penalty_val_accs = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E2_penalty_low-high_low-high_val_accs.npy")
)
single_eegs_E2_penalty_weights = np.load(
    os.path.join(RESULTS_DIR, "single_eegs_E2_penalty_low-high_low-high_weights.npy")
)
```

```python
filename = "../images/individual_weights.mat"
savemat(
    filename,
    {
        "single_eegs_E1_weights": single_eegs_E1_weights,
        "single_eegs_E2_baseline_weights": single_eegs_E2_baseline_weights,
        "single_eegs_E2_penalty_weights": single_eegs_E2_penalty_weights,
    },
)
single_eegs_E1_weights.shape
```

```python
fig, ax = plt.subplots()

plot_accuracy(single_eegs_E1_train_accs, ax, "")
plot_accuracy(
    single_eegs_E1_val_accs, ax, "E1 Low vs High Individual Mean Accuracy vs Time"
)

colors = ["C0", "C1"]
lines = [Line2D([0], [0], color=c) for c in colors]
labels = ["Train", "Validation"]
plt.legend(lines, labels)

plt.tight_layout()

plt.savefig("../images/E1_Individual.eps", format="eps")
plt.savefig("../images/E1_Individual")

plt.show()
```
